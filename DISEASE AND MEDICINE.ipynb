```python
# ==============================================================
#       MEDICINE SHOP ASSISTANT & DOCTOR APPOINTMENT SYSTEM
# ==============================================================
# Python Mini Project
#
# Features:
# 1. Patient Registration
# 2. Patient Details
# 3. Symptom Checking
# 4. Medicine Information
# 5. Severe Case Detection
# 6. Doctor Information
# 7. Doctor Appointment
# 8. Medicine Categories
# 9. First-Aid Information
# 10. Medicine Shop Bill
# 11. Medicine Stock
# 12. Search Condition
# 13. Patient Summary
#
# IMPORTANT:
# This is an educational project.
# It does not diagnose diseases or prescribe medicines.
# For serious symptoms, consult a qualified doctor.
# ==============================================================


# ==============================================================
# 1. MEDICINE DATABASE
# ==============================================================

medicines = {

    "fever": [
        "Paracetamol - commonly used for fever and pain",
        "ORS - useful when dehydration is present"
    ],

    "cough": [
        "Cough lozenges",
        "Honey-based cough preparation",
        "Cough syrup - type depends on the cough"
    ],

    "cold": [
        "Saline nasal spray",
        "Warm fluids",
        "Steam inhalation",
        "Cold-relief products - ask a pharmacist"
    ],

    "headache": [
        "Paracetamol - commonly used for mild pain",
        "Adequate water and rest"
    ],

    "stomach pain": [
        "ORS if vomiting or diarrhea is present",
        "Antacid products for some acidity-related symptoms",
        "Medical advice if pain is severe or persistent"
    ],

    "leg pain": [
        "Rest",
        "Gentle stretching when appropriate",
        "Paracetamol may help some mild pain"
    ],

    "body pain": [
        "Rest",
        "Adequate fluids",
        "Paracetamol may help some mild pain"
    ],

    "sore throat": [
        "Throat lozenges",
        "Warm salt-water gargle",
        "Warm fluids"
    ],

    "acidity": [
        "Antacid products",
        "Avoid foods that trigger acidity",
        "Medical advice if symptoms repeatedly occur"
    ],

    "indigestion": [
        "Antacid products",
        "Small meals",
        "Adequate water"
    ],

    "diarrhea": [
        "ORS",
        "Adequate fluids",
        "Medical advice if severe or persistent"
    ],

    "vomiting": [
        "ORS in small frequent amounts if tolerated",
        "Adequate fluids",
        "Medical advice if severe or persistent"
    ],

    "constipation": [
        "Adequate water",
        "Fiber-rich food",
        "Ask a pharmacist before using laxatives"
    ],

    "allergy": [
        "Antihistamine products may help some allergies",
        "Avoid known allergens",
        "Medical advice if symptoms are severe"
    ],

    "skin itching": [
        "Moisturizer",
        "Calamine-type lotion",
        "Ask a pharmacist about antihistamine products"
    ],

    "minor cut": [
        "Antiseptic suitable for minor wounds",
        "Sterile dressing",
        "Bandage"
    ],

    "minor burn": [
        "Cool running water",
        "Sterile non-stick dressing",
        "Medical attention for serious burns"
    ],

    "muscle pain": [
        "Rest",
        "Gentle movement when appropriate",
        "Paracetamol may help some mild pain"
    ]
}


# ==============================================================
# 2. CONDITION INFORMATION DATABASE
# ==============================================================

condition_information = {

    "fever":
        "Fever can occur with infections and many other conditions.",

    "cough":
        "Cough may occur due to infections, allergies, asthma or irritation.",

    "cold":
        "Common cold may cause sneezing, runny nose and nasal congestion.",

    "headache":
        "Headache can have many causes including stress and dehydration.",

    "stomach pain":
        "Stomach pain can have many causes and sometimes requires examination.",

    "leg pain":
        "Leg pain may occur because of muscle strain, injury or other causes.",

    "body pain":
        "Body pain can occur during infections or after physical activity.",

    "sore throat":
        "Sore throat may occur due to infections or irritation.",

    "acidity":
        "Acidity may cause burning discomfort and may be related to reflux.",

    "indigestion":
        "Indigestion can cause discomfort, fullness and bloating.",

    "diarrhea":
        "Diarrhea can result in loss of fluids and electrolytes.",

    "vomiting":
        "Repeated vomiting may cause dehydration.",

    "constipation":
        "Constipation may involve difficult or infrequent bowel movements.",

    "allergy":
        "Allergies may cause sneezing, itching, rashes or other symptoms.",

    "skin itching":
        "Skin itching can occur because of allergies, irritation or skin conditions.",

    "minor cut":
        "Minor cuts should be cleaned and covered with an appropriate dressing.",

    "minor burn":
        "Minor burns should be cooled with cool running water.",

    "muscle pain":
        "Muscle pain may occur after exercise, strain or minor injury."
}


# ==============================================================
# 3. SEVERE SYMPTOMS
# ==============================================================

severe_symptoms = [

    "chest pain",
    "difficulty breathing",
    "breathing difficulty",
    "severe bleeding",
    "heavy bleeding",
    "unconscious",
    "unconsciousness",
    "seizure",
    "stroke",
    "severe abdominal pain",
    "very high fever",
    "severe headache",
    "face swelling",
    "swelling of face",
    "blood vomiting",
    "vomiting blood",
    "black stool",
    "serious injury",
    "major injury"
]


# ==============================================================
# 4. DOCTOR DATABASE
# ==============================================================

doctors = [

    {
        "name": "Dr. Ananya Sharma",
        "speciality": "General Physician",
        "phone": "9876543210",
        "time": "10:00 AM - 1:00 PM",
        "days": "Monday to Friday"
    },

    {
        "name": "Dr. Rahul Verma",
        "speciality": "General Physician",
        "phone": "9876501234",
        "time": "4:00 PM - 7:00 PM",
        "days": "Monday to Saturday"
    },

    {
        "name": "Dr. Priya Singh",
        "speciality": "Internal Medicine",
        "phone": "9876512345",
        "time": "11:00 AM - 2:00 PM",
        "days": "Tuesday to Saturday"
    },

    {
        "name": "Dr. Arjun Mehta",
        "speciality": "Family Physician",
        "phone": "9876523456",
        "time": "5:00 PM - 8:00 PM",
        "days": "Monday to Friday"
    }
]


# ==============================================================
# 5. MEDICINE SHOP INVENTORY
# ==============================================================

inventory = {

    "Paracetamol": {
        "price": 30,
        "stock": 100
    },

    "ORS": {
        "price": 25,
        "stock": 80
    },

    "Cough Lozenges": {
        "price": 50,
        "stock": 60
    },

    "Saline Nasal Spray": {
        "price": 90,
        "stock": 40
    },

    "Antacid": {
        "price": 80,
        "stock": 50
    },

    "Bandage": {
        "price": 20,
        "stock": 100
    },

    "Antiseptic": {
        "price": 70,
        "stock": 50
    },

    "Calamine Lotion": {
        "price": 75,
        "stock": 40
    }
}


# ==============================================================
# 6. PATIENT REGISTRATION
# ==============================================================

def register_patient():

    print("\n")
    print("=" * 70)
    print("              PATIENT REGISTRATION")
    print("=" * 70)

    print("\nPlease enter the patient's details.")

    name = input("Patient Name: ")

    while True:

        try:

            age = int(input("Patient Age: "))

            if age < 0 or age > 120:

                print("Please enter a valid age.")

            else:

                break

        except ValueError:

            print("Please enter age as a number.")

    gender = input("Gender: ")

    phone = input("Phone Number: ")

    address = input("Address: ")

    blood_group = input("Blood Group (optional): ")

    emergency_contact = input("Emergency Contact Number: ")

    patient = {

        "name": name,
        "age": age,
        "gender": gender,
        "phone": phone,
        "address": address,
        "blood_group": blood_group,
        "emergency_contact": emergency_contact
    }

    print("\n")
    print("=" * 70)
    print("             REGISTRATION SUCCESSFUL")
    print("=" * 70)

    print("Patient Name:", name)
    print("Age:", age)
    print("Gender:", gender)
    print("Phone:", phone)

    return patient


# ==============================================================
# 7. DISPLAY PATIENT DETAILS
# ==============================================================

def display_patient_details(patient):

    print("\n")
    print("=" * 70)
    print("                 PATIENT DETAILS")
    print("=" * 70)

    print("Name              :", patient["name"])
    print("Age               :", patient["age"])
    print("Gender            :", patient["gender"])
    print("Phone             :", patient["phone"])
    print("Address           :", patient["address"])
    print("Blood Group       :", patient["blood_group"])
    print("Emergency Contact :", patient["emergency_contact"])


# ==============================================================
# 8. DISPLAY HEADER
# ==============================================================

def display_header(patient):

    print("\n")
    print("=" * 70)
    print("          MEDICINE SHOP ASSISTANT SYSTEM")
    print("=" * 70)

    print("Patient:", patient["name"],
          "| Age:", patient["age"])

    print("=" * 70)


# ==============================================================
# 9. DISPLAY CONDITIONS
# ==============================================================

def display_conditions():

    print("\n")
    print("=" * 70)
    print("                  COMMON CONDITIONS")
    print("=" * 70)

    for number, condition in enumerate(condition_information, start=1):

        print(number, ".", condition.title())


# ==============================================================
# 10. CHECK SEVERE CASE
# ==============================================================

def check_severe_case(symptoms):

    symptoms = symptoms.lower()

    for severe in severe_symptoms:

        if severe in symptoms:

            return True

    return False


# ==============================================================
# 11. FIND MEDICINES
# ==============================================================

def find_medicines(symptoms):

    found_medicines = []

    symptoms = symptoms.lower()

    for condition in medicines:

        if condition in symptoms:

            for medicine in medicines[condition]:

                if medicine not in found_medicines:

                    found_medicines.append(medicine)

    return found_medicines


# ==============================================================
# 12. FIND CONDITION INFORMATION
# ==============================================================

def find_condition_information(symptoms):

    information = []

    symptoms = symptoms.lower()

    for condition in condition_information:

        if condition in symptoms:

            information.append(
                condition_information[condition]
            )

    return information


# ==============================================================
# 13. CUSTOMER SYMPTOM CHECK
# ==============================================================

def symptom_check(patient):

    print("\n")
    print("=" * 70)
    print("                  SYMPTOM CHECK")
    print("=" * 70)

    print("Patient:", patient["name"])
    print("Age:", patient["age"])

    print("\nExamples:")
    print("fever")
    print("fever and cough")
    print("fever cough cold")
    print("headache")
    print("stomach pain")
    print("leg pain")
    print("cough and cold")

    symptoms = input("\nEnter patient's symptoms: ")

    # ----------------------------------------------------------
    # Severe case check
    # ----------------------------------------------------------

    if check_severe_case(symptoms):

        print("\n")
        print("!" * 70)
        print("                 POSSIBLE SEVERE CASE")
        print("!" * 70)

        print("\nThe symptoms entered may require urgent medical evaluation.")

        print("\nPlease contact emergency medical services or")
        print("visit the nearest emergency department when appropriate.")

        choice = input(
            "\nWould you like to see doctor details? (yes/no): "
        )

        if choice.lower() == "yes":

            display_doctors()

        return


    # ----------------------------------------------------------
    # Find medicine information
    # ----------------------------------------------------------

    found_medicines = find_medicines(symptoms)

    information = find_condition_information(symptoms)

    print("\n")
    print("=" * 70)
    print("                   RESULT")
    print("=" * 70)

    if not found_medicines:

        print("No matching condition was found.")

        print("\nPlease consult a qualified healthcare professional.")

        return

    print("Patient:", patient["name"])

    print("\nGeneral Information:")

    for item in information:

        print("-", item)

    print("\nCommonly Used Products / Supportive Measures:")

    for number, medicine in enumerate(found_medicines, start=1):

        print(number, ".", medicine)

    print("\nIMPORTANT:")
    print("These are general information only.")
    print("They are NOT a personalized prescription.")
    print("Always check labels and seek professional advice when needed.")


# ==============================================================
# 14. SEARCH CONDITION
# ==============================================================

def search_condition():

    print("\n")
    print("=" * 70)
    print("                 SEARCH CONDITION")
    print("=" * 70)

    search = input("Enter condition: ").lower()

    if search in condition_information:

        print("\nCondition:", search.title())

        print(
            "\nInformation:",
            condition_information[search]
        )

        print("\nRelated products/supportive measures:")

        for medicine in medicines[search]:

            print("-", medicine)

    else:

        print("\nCondition not found.")

        print("Please consult a qualified healthcare professional.")


# ==============================================================
# 15. DISPLAY DOCTORS
# ==============================================================

def display_doctors():

    print("\n")
    print("=" * 70)
    print("                    DOCTORS")
    print("=" * 70)

    for number, doctor in enumerate(doctors, start=1):

        print("\nDoctor", number)

        print("Name       :", doctor["name"])

        print("Speciality :", doctor["speciality"])

        print("Phone      :", doctor["phone"])

        print("Available  :", doctor["time"])

        print("Days       :", doctor["days"])


# ==============================================================
# 16. BOOK APPOINTMENT
# ==============================================================

def book_appointment(patient):

    display_doctors()

    print("\n")

    try:

        choice = int(input("Select doctor number: "))

        if choice < 1 or choice > len(doctors):

            print("Invalid doctor number.")

            return

    except ValueError:

        print("Please enter a valid number.")

        return

    doctor = doctors[choice - 1]

    print("\nSelected Doctor:")

    print("Name:", doctor["name"])

    print("Speciality:", doctor["speciality"])

    print("Available Time:", doctor["time"])

    date = input("Enter appointment date: ")

    print("\n")
    print("=" * 70)
    print("               APPOINTMENT CONFIRMED")
    print("=" * 70)

    print("Patient Name :", patient["name"])

    print("Patient Age  :", patient["age"])

    print("Doctor       :", doctor["name"])

    print("Speciality   :", doctor["speciality"])

    print("Date         :", date)

    print("Time         :", doctor["time"])

    print("Doctor Phone :", doctor["phone"])

    print("\nPlease arrive on time.")


# ==============================================================
# 17. MEDICINE CATEGORIES
# ==============================================================

def medicine_categories():

    categories = [

        "Pain and fever relief",
        "Cold and cough products",
        "Digestive health products",
        "Allergy products",
        "Oral rehydration products",
        "First-aid products",
        "Skin-care products",
        "Throat-care products"
    ]

    print("\n")
    print("=" * 70)
    print("                MEDICINE CATEGORIES")
    print("=" * 70)

    for number, category in enumerate(categories, start=1):

        print(number, ".", category)


# ==============================================================
# 18. FIRST-AID INFORMATION
# ==============================================================

def first_aid():

    print("\n")
    print("=" * 70)
    print("                 FIRST-AID INFORMATION")
    print("=" * 70)

    print("\n1. Minor Cut")

    print("Clean the wound and cover it with a sterile dressing.")

    print("\n2. Minor Burn")

    print("Cool the affected area with cool running water.")

    print("\n3. Fever")

    print("Monitor temperature, rest and maintain adequate fluids.")

    print("\n4. Dehydration")

    print("ORS can help replace fluids and electrolytes.")

    print("\n5. Serious Injury")

    print("Seek immediate professional medical assistance.")


# ==============================================================
# 19. DISPLAY INVENTORY
# ==============================================================

def display_inventory():

    print("\n")
    print("=" * 70)
    print("                 MEDICINE INVENTORY")
    print("=" * 70)

    print(
        f"{'Medicine':25} {'Price':10} {'Stock':10}"
    )

    print("-" * 50)

    for medicine, details in inventory.items():

        print(
            f"{medicine:25} "
            f"₹{details['price']:<9} "
            f"{details['stock']:<10}"
        )


# ==============================================================
# 20. CREATE MEDICINE BILL
# ==============================================================

def create_bill(patient):

    print("\n")
    print("=" * 70)
    print("                  MEDICINE BILL")
    print("=" * 70)

    display_inventory()

    cart = []

    while True:

        medicine_name = input(
            "\nEnter medicine name or type 'done': "
        )

        if medicine_name.lower() == "done":

            break

        if medicine_name not in inventory:

            print("Medicine not available in inventory.")

            continue

        try:

            quantity = int(
                input("Enter quantity: ")
            )

            if quantity <= 0:

                print("Quantity must be positive.")

                continue

        except ValueError:

            print("Enter a valid quantity.")

            continue

        if quantity > inventory[medicine_name]["stock"]:

            print("Not enough stock available.")

            continue

        price = inventory[medicine_name]["price"]

        total = price * quantity

        cart.append(
            {
                "medicine": medicine_name,
                "quantity": quantity,
                "price": price,
                "total": total
            }
        )

        inventory[medicine_name]["stock"] -= quantity

        print("Item added successfully.")

    if len(cart) == 0:

        print("\nNo items were purchased.")

        return

    print("\n")
    print("=" * 70)
    print("                  FINAL BILL")
    print("=" * 70)

    print("Patient:", patient["name"])

    print("-" * 70)

    grand_total = 0

    for item in cart:

        print(
            item["medicine"],
            "x",
            item["quantity"],
            " = ₹",
            item["total"]
        )

        grand_total += item["total"]

    print("-" * 70)

    print("TOTAL AMOUNT: ₹", grand_total)

    print("=" * 70)

    print("Thank you for visiting our medicine shop.")


# ==============================================================
# 21. PATIENT SUMMARY
# ==============================================================

def patient_summary(patient):

    print("\n")
    print("=" * 70)
    print("                  PATIENT SUMMARY")
    print("=" * 70)

    print("Patient Name       :", patient["name"])

    print("Age                :", patient["age"])

    print("Gender             :", patient["gender"])

    print("Phone              :", patient["phone"])

    print("Address            :", patient["address"])

    print("Blood Group        :", patient["blood_group"])

    print("Emergency Contact  :",
          patient["emergency_contact"])

    print("\nThis information is for the current project session.")


# ==============================================================
# 22. ABOUT PROJECT
# ==============================================================

def about_project():

    print("\n")
    print("=" * 70)
    print("                    ABOUT PROJECT")
    print("=" * 70)

    print("\nProject Name:")
    print("Medicine Shop Assistant & Doctor Appointment System")

    print("\nProgramming Language:")
    print("Python")

    print("\nPython Concepts Used:")

    print("1. Variables")
    print("2. Lists")
    print("3. Dictionaries")
    print("4. Functions")
    print("5. Loops")
    print("6. Conditional Statements")
    print("7. String Handling")
    print("8. Exception Handling")
    print("9. User Input")
    print("10. Menu-driven Programming")

    print("\nProject Purpose:")

    print("The system demonstrates how Python can be used")
    print("to develop a simple medicine-shop management system.")

    print("\nDisclaimer:")

    print("This project provides general information only.")
    print("It does not diagnose diseases or prescribe medicines.")


# ==============================================================
# 23. MAIN MENU
# ==============================================================

def main_menu(patient):

    while True:

        display_header(patient)

        print("\n1. Check Symptoms")

        print("2. Search Condition")

        print("3. View Patient Details")

        print("4. View Common Conditions")

        print("5. Medicine Categories")

        print("6. First-Aid Information")

        print("7. View Doctors")

        print("8. Book Doctor Appointment")

        print("9. View Medicine Inventory")

        print("10. Create Medicine Bill")

        print("11. Patient Summary")

        print("12. About Project")

        print("13. Exit")

        print("\n")

        choice = input("Enter your choice: ")

        if choice == "1":

            symptom_check(patient)

        elif choice == "2":

            search_condition()

        elif choice == "3":

            display_patient_details(patient)

        elif choice == "4":

            display_conditions()

        elif choice == "5":

            medicine_categories()

        elif choice == "6":

            first_aid()

        elif choice == "7":

            display_doctors()

        elif choice == "8":

            book_appointment(patient)

        elif choice == "9":

            display_inventory()

        elif choice == "10":

            create_bill(patient)

        elif choice == "11":

            patient_summary(patient)

        elif choice == "12":

            about_project()

        elif choice == "13":

            print("\n")
            print("=" * 70)

            print("Thank you for using the Medicine Shop Assistant.")

            print("Please take care of your health.")

            print("=" * 70)

            break



        else:

            print("\nInvalid choice.")

            print("Please enter a number from 1 to 13.")

def start_program():

    print("\n")
    print("=" * 70)
    print("       WELCOME TO MEDICINE SHOP ASSISTANT SYSTEM")
    print("=" * 70)

    print("\nBefore using the system,")
    print("please register the patient.")

    patient = register_patient()

    print("\nStarting the Medicine Shop System...")

    main_menu(patient)

if __name__ == "__main__":

    start_program()
```
